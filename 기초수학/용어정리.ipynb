### 이처럼 한가지 결과를 두 가지로 표현하는 것이 베르누이 모형이다

### Likelihood(우도)
### 이미 관측한 데이터를 고정해 놓고, 파라미터에 따라 그데이터에 부여되는 확률을 비교하는 함수
### y = [1,0,1,1,0], p= 6 이면
### L(0.6) = 0.6 * 0.4 * 0.6 * 0.6 * 0.4 = 0.03456
### 이  경우 L(p)는 p를 넣었을 떄의 우도라고 읽을 수 있다
### 위에 계산식을 말로 풀어보면
### 다섯 개의 과제 제출 기록이 [1,0,1,1,0]이라고 하면 이 기록을 그대로 공동 제출 확률 두 개를 비교한다
### p = 0.6이면 실제 기록에 준 확률의 곱은 0.6 * 0.4 * 0.6 * 0.6 * 0.4 = 0.03456
### p = 0.8이면 실제 기록에 준 확률의 곱은 0.8 * 0.2 * 0.8 * 0.8 * 0.2 = 0.02048
### 제출 확률 80으로 설정해놓았다면 제출자에게는 유리하지만 미제출 두 명에게는 20%씩만 준다
### -> 왜 쓸까 -> 어느 모델 설정한 값이 지금 현재 기록에 더 잘 맞는지 숫자로 비교하고 싶다

### probability 와 Likelihood(우도)의 관점
### 두 개념은 확률식을 서로 다른 관점에서 바라본다
### probability는 모델을 정해두고 가능한 결과를 살펴보는 방법이라면
### Likelihood는 실제 출력된 결과를 보고 모델의 파라미터 값을 보겠다는 의미
### 확률 관점에서 p = 0.6일 때라고 가정하면 이 0.6이라는 숫자값이 중요하다 구매 확률이 60%야 라는 질문을 던진다면
### Likelihood(우도) 관점에서는 실제 구매, 미구매 기록을 고정한 상태에서 L(0.6),L(0.8)을 비교하겠다라는 의미
### 질문 - 수업 전에 이번 과제를 제출할 확률을 0.6으로 설정했다면 연준이 이번 과제를 제출할 확률은?=> 확률 관점
### 수업이 끝나고 난 뒤 [1,0,1,1,0]이라는 실제 기록을 확인한 뒤 P=0.6과 P=0.8 중 어느 설정이 이 기록에 더 큰 확률을 주는지? => 우도 관점

### 파라미터 후보와 후보 비교
### 파라미터의 후보라는 것은 시럼해볼 파라미터 값을 의미한다
### 각 후보에 같은 데이터를 넣어 우도를 구한 다음에 가장 큰 우도를 주는 후보를 선택한다
### p = [0.2, 0.4, 0.6, 0.8]의 우도는 순서대로 [0.005,0.023,0.034,0.020]이다
### 그렇다면 가장 큰 우도로 연결된 후보는 P=0.6이다
### 이 계산식을 말로 풀어보면
### [1,0,1,1,0]에 대한 제출 확률을 p = [0.2, 0.4, 0.6, 0.8] 산출하면 [0.005,0.023,0.034,0.020]다
### 따라서 제출 확률에 대한 우도값이 가장 큰 것은 P=0.6이기 떄문에 지금 [1,0,1,1,0] 기록에 더 잘맞는 설정값을 찾는 일(p=0.6)이다
### p가 무조건 높다고 좋은 것은 아님(ex. P=0.8 탈락)

### MLE
### 관측 데이터의 우도가 가장 커지는 파라미터를 찾는 방법
### 공통 p를 사용하는 독립 베르누이 모형에서는 MLE = 구매수 / 전체 고객 수
### 구매 3명 전체 5명이면 3/5 = 0.6 => 이 모형ㅇ에서는 0~1 사이에서 0.6이 우도를 가장 크게 만든다
### 예를 들어 예약 방문 기록이 [1,0,1,1,0]이라고 보면 방문 3명 미방문 2명이 모든 예약에 같은 방문 확률 P를 사용하겠다고 가정해보면
### 각 결과가 독립인 베르누이 모형에서는 우도를 가장 크게하는 P가 방문수 / 전체수인 3/5 = 0.6
### 따라서 0.6이 최대우도추정값이고 그때 전체 기록의 우도는 0.034 정도
### -> 사람이 임의로 정한 확률 대신에 관측 자료 결과에 근거하여 파라미터를 정하기 위해서 최대 우도 추정을 기준으로 사용한다

### 인덱스와 argmax
### 인덱스는 배열의 위치 번호, np.argmax()는 가장 큰 값 자체가 아니라 그 값이 들어있는 위치 번호를 반환
### 우도 배열 [0.005,0.023,0.034,0.020]이라고 가정해보면
### 가장 큰 우도는 0.034이며 순서는 세 번째에 있다는 것을 알기 위함
### -> 가장 좋은 평가값을 빠르게 찾고 그 뒤 값을 만든 설정까지 빠르게 찾기 위함

### 로그
### 어떤 밑을 제곱하면 이 값이 되는가를 알려주는 연산자
### 예를 들어 2를 3제곱하면 8이니까 밑이 2인 8의 로그는 3이다
### 그렇다면 우도랑 뭔 상관이냐
### 예약 세 건의 실제 결과에 확률이 0.8, 0.7, 0.7 이라고 가정을 해보면
### 우도는 이 세 값을 곱한 0.392 값을 의미한다
### 이 우도의 자연로그는 ln(0.392)로 구할 수 있고 ln(0.8)+ln(0.7)+ln(0.7)로 얻을 수 있다
### 즉 로그의 특성을 이용하면 확률을 모두 곱하는 과정을 확률의 로그를 더하는 과정으로 바꿀 수가 있다
### 기록이 매우 많아서 계속 곱해서 0에 한도 끝도 없이 가까워지면 구분하기 어려우니까
### 같은 비교지만 로그를 써서 안정적으로 계산하겠다는 의미다

### 로그우도(Log Likelihood)
### 우도에 자연로그를 적용한 값
### 각 관측 확률의 자연로그를 더해서도 같은 값을 얻는다
### 같은 데이터에서는 큰 로그우도를 선호한다
### q = [0.8, 0.7, 0.7]이면 L = 0.392고 ln(L)=lnln(0.8)+ln(0.7)+ln(0.7)=-0.936
### 코드에서는 np.sum(np.log(q))를 사용할 수 있다
### 예를 들어 다른 모델이 같은 세 결과에 각각 0.5를 주면 우도는 0.125
### 로그우도는 약 -2.0794값
### -0.9365 > -2.0794이기 떄문에 첫 모델을 선택하려고 한다
### 우도로 비교해도 0.392 > 0.125보다는 크기 때문에 같은 선택을 한다

### 같은 데이터에서의 모델 비교
### 같은 고객의 같은 실제 결과에 대해 각 모델이 부여한 우도를 비교하겠다
### 한 고객에게 더 좋은 확률을 준 모델이 다른 고객 또는 전체 데이터에서 항상 좋은 것은 아니다
### 실제값[1, 0, 1]에서 A의 관측확률 [0.8,0.6, 0.6]은 우도 0.288
### B의 관측 확률 [0.6, 0.8, 0.5]는 0.240 => 이 두 모델을 비교했을 땐 A의 우도 값이 높으니까
### A 모델을 선택하겠다


```python
#2장 3강

# 손실 함수와 Loss
# 모델의 예측을 실제 정답과 비교해 숫자로 평가하는 규칙
# 금액 예측은 실제 값과 예측값의 차이로 MSE를 구할 수 있다
# 구매 여부와 상품 종류는 실제 정답에 주어진 확률로  BCE 또는 CE를 구할 수 있다
# 문제 종류에 맞게 계산 규칙을 선택한다
# 예를 들어 배달 시간 예측 모델 두 개를 비교하겠다고 가정해보면
# 실제 시간은 [10,20,30]분이고 예측 시간은 [12,18,27]분이라면
# 각 주문에서 실제 시간과 예측 시간의 차이를 제곱하면[4,4,9]다 => 평균을 계산해보면 17/3 = 5.66
# 같은 주문에서 B의 평균 제곱 오차가 8.667이라면 5.66인 A 모델이 더 좋았다고 평가할 수 있다
# 손실함수는 모델의 예측을 비교하고 학습 중 어떤 것을 줄여야하는지 정하는 역할을 한다

# 제곱오차합과 SSE
# 제곱 오차를 모두 더한 값
# 제곱 오차 [4,4,9]의 SSE는 17이다
# 배달 3건에 대해서 A모델의 제곱 오차가 [4,4,9], B의 제곱 오차가 [1,16,9]라고 가정해보면
# A의 합계는 17, B는 26 => 이 합계가 의미하는 것이 SSE
# 같은 세 건에 대해서는 A의 전제 제곱 오차가 더 작다
# B가 비록 첫 주문을 A모델보다 잘 맞췄더라도 전체 합계에서는 A가 더 좋은 모델일 숭 ㅣㅆ다
# SSE / 전체 주문 건수 = MSE
# 이걸 왜 함?
# -> SSE를 통해 MSE를 구할 수 있고 합계가 커진 이유 자체가 오차인지 데이터의 수에 영향을 받았는지 구분할 수 있다

# 평균 제곱 오차 MSE
# 제곱오차의 평균
# 각 고객의 오차를 제곱하고 모두 더한 뒤 전체 고객 수로 나눈다
# 배달 세건의 실제 시간이 [10,20,30]분이고 예측이 [12,18,27]분이면
# 실체 예측은 [-2,2,3], 각 오차의 제곱은 [4,4,9] 합계 17을 3(전체 고객 수)로 나누면 5.66
# 다른 B모델의 MSE가 8.66이면 A 모델을 선택하는 것이 맞다
# 그렇다면 왜 중요하게 배움?
# 여러 수치의 예측 제곱 오차를 하나의 평균으로 요약해 비교하기 위해서
# 오차를 전체 수로 평균을 내기 때문엑 ㅏㄱ 한 건에 대한 오차 범위를 대략적으로 파악할 수 있음

# MSE의 단위
# 오차를 제곱하면 단위도 제곱하면 된다
# 단위가 만원단위였다면 제곱오차에 대한 해석도 (만원)^2 단위로 읽으면 된다

# 이진 교차엔트로피와 BCE
# Binary Cross Entropy, 구매 / 미구매 두가지 결과를 구분하는 분류 손실 함수
# 한 고객의 BCE는 실제 정답 확률에 음의 로그를 적용한 값
# 예를 들어보면
# 두 예약자에 실제 방문 결과가 [1,0]이고 A의 방문 확률이 [0.8, 0.3]이라고 가정하면
# 첫번째 손님의 손실은 -ln(0.8)= 약 0.223, 둘째는 미방문 이니까 -ln(0.7) = 약 0.356 정도
# 평균 BCE 값은 약 0.289 정도
# B가 두 사람 모두에게 방문 확률 0.5를 주었다면 두 손실은 각각 약  0.6931로 평균도 0.6931로 나타난다
# 같은 두 예약 확률에서는 BCE가 작은 A모델을 선택한다
# 왜 쓰냐
# 두 예측 결과를 학습하고 평가할 기준으로 쓴다 각 정답에 어느 정도의 확률을 부여헀는지 알 수 있다

# 범주형 교차엔트로피와 CE 
# Categorical Cross Entropy, 여러 종류 중 실제 정답이 하나인 분류에 쓰이는 손실
# 상품 B의 정답이 수박 [배, 수박, 사과] 확률이 [0.2, 0.6, 0.2] 라면 CE = - ln(0.6) = 약 0.511
# 예를 들어 과일 종류를 [배, 수박, 사과]순서로 분류한다고 했을 때
# 첫 문의의 실제 정답은 수박이고 예측 확률이 [0.2, 0.6, 0.2] 일때 정답인 수박의 0.6을 선택해
# -ln(0.6)=약 0.511을 계산한다
# 두번째의 정답 확률이 0.8이라면 CE = 0.2231 두 과일 선택의 평균을 내보면 0.367 정도
# 왜 쓰느냐 과일 선택처럼 대상마다 열의 확률을 고르고 대상별 손실을 평균 내는 계산을 활용할 수 있다

# 예측 클래스와 손실의 차이
# 예측 클래스는 모델이 최종적으로 선택한 종류
# 손실은 실제 정답에 준 확률을 평가한 값
# 정답을 맞춘다고 손실이 꼭 0은 아님
# 예를 들어
# 실제 구매 고객에게 확률 0.9와 0.6을 주면 임계값 0.5에서 둘 다 구매로 예측
# 하지만 BCE는 약 0.105와 0.511로 다를 수 있음
# 즉 -> 정답을 맞췄다는 사실만으로 예측의 품질을 알 수가 없기 때문에 손실 점수를 확인해야 한다
# 두 가지 사실, 정답을 맞췄는가 / 확률 예측 자체가 얼마나 좋은가?에 대한 답을 할 수 있다

# 정규분포(가우시안)
# 평균을 중심으로 좌우가 대칭인 종 모양의 연속형 분포
# 예를 들어보면 배달 모델에서 주문 도착 시간을 20분으로 예상했다고 가정하면
# 실제 배달 시간이 19분 21분처럼 약간의 오차가 발생할 수는 있다
# 비슷한 조건에서 주문에서 작은 차이는 자주 큰 차이는 드물게 나타난다고 설명하기 위해서
# 정규분포를 가정할 수 있다
```
