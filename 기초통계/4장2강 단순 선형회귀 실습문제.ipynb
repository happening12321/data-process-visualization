### 문제 1-2. 기울기와 절편을 직접 계산하기

최소제곱법 공식을 이용해 회귀계수를 직접 계산합니다.

```text
β1 = Σ(x - x̄)(y - ȳ) / Σ(x - x̄)²

β0 = ȳ - β1x̄
```

**요구사항**

1. `sample_df`의 `distance_km`를 x로 저장하세요.
2. `delivery_cost`를 y로 저장하세요.
3. 위 공식을 이용해 기울기 `beta1`을 계산하세요.
4. 절편 `beta0`를 계산하세요.
5. 다음 형태의 회귀식을 출력하세요.

```text
delivery_cost = beta0 + beta1 × distance_km
```

6. 기울기와 절편의 의미를 각각 한 문장으로 설명하세요.

````python
# 문제 1-2 코드를 작성하세요.

x=sample_df['distance_km']
y=sample_df['delivery_cost'] # y=sample_df['delivery_cost'].to_numpy() # 컴퓨터가 계산하는 식으로 만듦

x_mean = x.mean()
y_mean = y.mean()

bata1 = np.sum((x - x_mean) * (y - y_mean)) / np.sum((x - x_mean) ** 2)
bata0 = y_mean - bata1 * x_mean

print("기울기: ", bata0)
print("절편: ", bata1)
print("회귀식: delivery_cost = "f"{bata0:.2f} + "f"{bata1:.2f} * distance_km")

````
