## 필수 1. 배송 상태의 상대도수 확률과 기본 법칙

### 문제 1-1. 배송 상태별 상대도수 확률 계산하기

**요구사항**

1. `delivery_status`의 각 상태별 빈도수를 계산하세요.
2. `value_counts(normalize=True)`를 이용해 상태별 상대도수 확률을 계산하세요.
3. 확률을 소수점 4자리로 출력하세요.
4. 가장 높은 확률을 가진 배송 상태와 확률을 출력하세요.

```python
# 문제 1-1 코드를 작성하세요.

# 빈도수
status_count = df['delivery_status'].value_counts()

# 상대도수
status_prob = df['delivery_status'].value_counts(normalize=True).round(4)

# 출력
print("배송 상태별 빈도수", status_count)
print("\n배송 상태별 상대도수", status_prob)

# 가장 높은 확률
most_common_status = status_prob.idxmax()
highest_prob = status_prob.max()

print(f"\n가장 높은 확률을 가진 배송 상태: {most_common_status}")
print(f"확률: {highest_prob:.4f}")
```
