```python
import pandas as pd

df = pd.DataFrame({
    '이름': ['Alice', 'Bob', 'Charlie', None],
    '나이': [25, None, 30, 28],
    '점수': [90, 85, None, 92]
})

# 결측치가 있는 모든 행 삭제
df_dropped = df.dropna()
print(df_dropped)

# '나이' 컬럼에 결측치가 있는 행만 삭제
df_dropped_age = df.dropna(subset=['나이'])
print(df_dropped_age)
```

```python
# 평균으로 결측치 대체
df['나이'] = df['나이'].fillna(df['나이'].mean())

# 중앙값으로 결측치 대체
df['점수'] = df['점수'].fillna(df['점수'].median())

# 최빈값으로 결측치 대체 (범주형)
df['이름'] = df['이름'].fillna(df['이름'].mode()[0])

print(df)
```

```python
import pandas as pd

df = pd.DataFrame({
    "나이": [25, 30, None, 35, 40],
    "구매금액": [30000, 50000, None, 70000, 800000],
    "회원등급": ["일반", "VIP", "일반", None, "일반"]
})

# [결측치 처리 근거]
# 나이: 이상치가 없고 값의 분포가 크게 치우치지 않음 → 평균 대체
# 구매금액: 큰 금액의 이상치가 존재함 → 중앙값 대체
# 회원등급: 범주형 데이터이며 '일반'이 가장 많이 등장함 → 최빈값 대체

df["나이"] = df["나이"].fillna(df["나이"].mean())
df["구매금액"] = df["구매금액"].fillna(df["구매금액"].median())
df["회원등급"] = df["회원등급"].fillna(df["회원등급"].mode()[0])

print(df)
```

```python
import pandas as pd

df = pd.read_csv('./ecommerce_orders_online.csv')

Q1 = df['revenue'].quantile(0.25)
Q3 = df['revenue'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df['revenue'] < lower_bound) |
    (df['revenue'] > upper_bound)
]

print('하한값:', lower_bound)
print('상한값:', upper_bound)
print('이상치 개수:', len(outliers))

print(
    outliers[
        ['order_id', 'product_category', 'revenue']
    ].head()
)
```

```python
import pandas as pd
import numpy as np
from scipy import stats

df = pd.DataFrame({
    "상담시간": [
        8, 9, 10, 11, 10, 9, 12, 10, 11, 9,
        10, 8, 11, 10, 9, 12, 10, 11, 9, 10,
        8, 10, 11, 9, 10, 12, 9, 11, 10, 60
    ]
})

z_scores = np.abs(stats.zscore(df["상담시간"]))

outliers = df[z_scores > 3]

print(outliers)
```

```python
from scipy import stats

df = pd.DataFrame({'값': [10,12,11,13,9,11,12,10,8,10,11,9,12,13,10,11,9,10,12,11,10,9,11,12,10,300]})

z_scores = np.abs(stats.zscore(df['값']))

# |z| > 3인 이상치 탐지
outliers_z = df[z_scores > 3]
print(f'이상치 (Z-score > 3):\n{outliers_z}')
```

```python
import pandas as pd
import numpy as np
from scipy import stats

df = pd.read_csv('./customer_order_summary.csv')

order_value = df['avg_order_value'].dropna()

z_scores = np.abs(stats.zscore(order_value))

outliers = df.loc[order_value.index[z_scores > 3]]

print('이상치 개수:', len(outliers))

print(outliers[['customer_id','total_orders','avg_order_value']].head())
```

```python
import pandas as pd

df = pd.DataFrame({
    "가격": [10000, 12000, 13000, 15000, 16000, 17000, 18000, 20000, 1000000]
})

Q1 = df["가격"].quantile(0.25)
Q3 = df["가격"].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# 1. 제거
df_drop = df[
    (df["가격"] >= lower_bound) &
    (df["가격"] <= upper_bound)
]

# 2. Winsorizing
df["가격_winsorized"] = df["가격"].clip(
    lower=lower_bound,
    upper=upper_bound
)

print(f"상한값: {upper_bound:.0f}")
print(df.tail(2))
```
