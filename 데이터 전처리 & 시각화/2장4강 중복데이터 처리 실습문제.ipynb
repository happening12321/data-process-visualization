원본 `appointments`는 수정하지 않습니다.

### 중복 판단 시 핵심

```text
AppointmentID 반복
→ 동일 예약이 중복 기록되었을 가능성

PatientId 반복
→ 같은 환자가 여러 번 예약했을 가능성
→ 정상 반복일 수 있음
```

## 실습 준비

````python
import pandas as pd

# 원본 데이터 불러오기
appointments = pd.read_csv("Medical_Appoingtment_No_Shows.csv")

# 원본은 보존하고 실습용 복사본 생성
base = appointments.copy()

# 실습용 완전 중복 행 5개 생성
duplicate_rows = base.iloc[:5].copy()

# 완전 중복이 뒤에 추가되도록 구성
dirty = pd.concat(
    [base, duplicate_rows],
    ignore_index=True
)

print("원본 행 수:", len(appointments))
print("실습용 행 수:", len(dirty))
````
