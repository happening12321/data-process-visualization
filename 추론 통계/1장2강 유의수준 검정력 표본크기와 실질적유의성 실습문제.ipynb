### 문제 1-1. 중앙 냉방시설 비교 연구에 필요한 표본 수 설계

#### 문제 설명

중앙 냉방시설이 있는 주택과 없는 주택의 판매가격을 비교하는 연구를 준비하고 있습니다. 사전 조사에서 표준화 효과 크기는 `0.3` 정도로 예상했습니다.

다음 조건에서 그룹당 표본 크기에 따른 검정력을 비교하고, 목표 검정력 0.8을 확보하기 위해 필요한 표본 수를 계산하세요.

- 예상 효과 크기: `0.3`
- 유의수준: `0.05`
- 양측검정
- 비교할 그룹당 표본 수: `30명`, `100명`, `300명`

#### 요구사항

1. `TTestIndPower()` 객체를 생성하세요.
2. 그룹당 표본 수가 30명, 100명, 300명일 때의 검정력을 각각 계산하세요.
3. 표본 크기별 검정력을 소수점 셋째 자리까지 출력하세요.
4. 유의수준 0.05, 목표 검정력 0.8, 효과 크기 0.3일 때 그룹당 필요한 표본 수를 계산하세요.
5. 계산된 표본 수는 소수점 이하를 올림하여 정수로 출력하세요.
6. 세 표본 크기 중 목표 검정력 0.8을 충족하는 경우를 확인하세요.

#### 해석 질문

**Q1.** 효과 크기와 유의수준이 같을 때 표본 크기가 커지면 검정력은 어떻게 변하나요?  
**Q2.** 검정력이 낮으면 실제 차이가 존재할 때 어떤 오류의 위험이 커지나요?  
**Q3.** 필요한 표본 수를 소수점 이하 올림으로 처리하는 이유는 무엇인가요?  
**Q4.** 표본 수를 무조건 크게 설정하는 것이 항상 최선인가요?

#### 제출 결과

- 표본 크기별 검정력
- 목표 검정력에 필요한 그룹당 표본 수
- 목표 검정력 충족 여부
- Q1~Q4 답변

```python
# 필수 1 코드를 작성하세요.

import math
from statsmodels.stats.power import TTestIndPower

# 1. TTestIndPower 객체 생성
power_analysis = TTestIndPower()

# 기본 조건
effect_size = 0.3
alpha = 0.05

# 2. 그룹당 표본 수별 검정력 계산
sample_sizes = [30, 100, 300]

for n in sample_sizes:
    power = power_analysis.solve_power(
        effect_size=effect_size,
        nobs1=n,
        alpha=alpha,
        ratio=1,
        alternative="two-sided"
    )
    print(f"그룹당 표본 수 {n}명 → 검정력: {power:.3f}")

# 3. 목표 검정력 0.8을 위한 필요한 표본 수 계산
required_n = power_analysis.solve_power(
    effect_size=effect_size,
    power=0.8,
    alpha=alpha,
    ratio=1,
    alternative="two-sided"
)

# 4. 소수점 이하 올림
required_n_ceil = math.ceil(required_n)

print(f"\n목표 검정력 0.8을 위한 필요 표본 수: {required_n:.3f}")
print(f"올림한 그룹당 필요 표본 수: {required_n_ceil}명")

# 5. 목표 검정력 0.8 충족 여부
print("\n[목표 검정력 0.8 충족 여부]")

for n in sample_sizes:
    power = power_analysis.solve_power(
        effect_size=effect_size,
        nobs1=n,
        alpha=alpha,
        ratio=1,
        alternative="two-sided"
    )
    
    result = "충족" if power >= 0.8 else "미충족"
    print(f"그룹당 {n}명 → {result}")
```

### 필수 1 답변 작성란

- **Q1.** 효과 크기와 유의수준이 같을 때 표본 크기가 커지면 검정력은 어떻게 변하나요?

- 표본 크기가 커질수록 검정력이 증가합니다.
- 즉, 실제로 차이가 존재할 때 그 차이를 통계적으로 발견할 가능성이 높아집니다.

- **Q2.** 검정력이 낮으면 실제 차이가 존재할 때 어떤 오류의 위험이 커지나요?

- 제2종 오류(Type II error)의 위험이 커집니다.
- 즉, 실제로 차이가 존재하지만 이를 발견하지 못하고 귀무가설을 기각하지 못하는 오류입니다.

- **Q3.** 필요한 표본 수를 소수점 이하 올림으로 처리하는 이유는 무엇인가요?

- 표본 수는 사람이나 주택 등의 개체 수이므로 소수로 측정할 수 없기 때문입니다. 또한 필요한 최소 표본 수보다 부족해지지 않도록 올림합니다.

- 예를 들어 계산 결과가 87.5명이라면 87명으로 내림하면 목표 검정력을 확보하지 못할 수 있으므로 88명으로 올림합니다.

- **Q4.** 표본 수를 무조건 크게 설정하는 것이 항상 최선인가요?

- 아닙니다. 표본 수가 커지면 검정력은 높아지지만 조사 비용, 시간, 데이터 수집 노력도 증가합니다. 또한 필요 이상으로 큰 표본을 사용하면 자원 낭비가 발생할 수 있습니다.

- 따라서 연구에서는 일반적으로 목표 검정력(예: 0.8)을 확보하면서도 현실적으로 감당할 수 있는 적절한 표본 크기를 설정하는 것이 좋습니다.

#### Cohen’s d 공식

두 집단의 평균 차이를 합동 표준편차로 나누어 계산합니다.

$$
d = \frac{\bar{x}_1 - \bar{x}_2}{s_p}
$$

$$
s_p = \sqrt{\frac{(n_1-1)s_1^2 + (n_2-1)s_2^2}{n_1+n_2-2}}
$$

- $\bar{x}_1$, $\bar{x}_2$: 각 집단의 평균
- $s_1$, $s_2$: 각 집단의 표본 표준편차
- $n_1$, $n_2$: 각 집단의 표본 수
- $s_p$: 합동 표준편차
