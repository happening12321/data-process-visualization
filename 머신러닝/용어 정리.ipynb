```python
# x
# df
# feature : 입력 정보(컬럼), x
# label : 정답(컬럼), y
# train_test_split(test_size=0.2) : 학습용 데이터와 테스트용 데이터를 나누는 함수
# random_state=42 : 랜덤 시드값, 동일한 데이터셋을 나누기 위해 고정

# 선형 회귀 : 입력값들의 가중합으로 연속적인 숫자를 예측하는 모델
# - > 의료비 : 나이 + bmi + 흡연 여부

# 가중치(weight) : 입력값에 곱해지는 값, 학습을 통해 결정됨
# bias(편향 or 절편) : 모든 feature의 기여분을 더하기 전에 (기본 값 입력값과 상관없이 항상 더해지는 값, 학습을 통해 결정됨)
# - > 택시비의 기본 요금

# mae : 실제값과 예측값 사이의 절댓값을 평균 낸 값 (평균 절대 오차, 예측값과 실제값의 차이의 절댓값을 평균한 값)
# -> mae = (|실제값 - 예측값|) / n -> 오차가 2,-3, 4 이면 mae = (|2| + |-3| + |4|) / 3 = 3

# mse : 실제값과 예측값 사이의 제곱을 평균 낸 값 (평균 제곱 오차, 예측값과 실제값의 차이의 제곱을 평균한 값)
# -> mse = ((실제값 - 예측값)^2) / n -> 오차가 2,-3, 4 이면 mse = ((2^2) + (-3^2) + (4^2)) / 3 = 11.6667

# rmse : 실제값과 예측값 사이의 제곱을 평균 낸 값에 루트를 씌운 값 
# (평균 제곱근 오차, 예측값과 실제값의 차이의 제곱을 평균한 값에 루트를 씌운 값)
# -> rmse = sqrt(((실제값 - 예측값)^2) / n 

# 최소제곱법 : 예측값과 실제값 차이의 제곱합이 가장 작아지는 가중치와 편향을 계산

# 경사하강법 : 선형회귀의 가중치와 편향을 계산하는 방법 중 하나, loss가 최소가 되는 방향으로 가중치와 편향을 조금씩 수정
# 가중치와 편향을 임의로 초기화하고, 예측값과 실제값의 차이를 계산하여 오차를 최소화하는 방향으로 가중치와 편향을 조금씩 수정

# Epoch : 학습데이터 읽을 횟수, 100 Epoch = 교재를 100번 읽음
# 경사하강법에서 가중치와 편향을 수정하는 횟수, 학습률이 너무 크면 loss가 발산하고, 학습률이 너무 작으면 loss가 수렴하지 않음

# learning rate : 한 번에 학습할 양, 한 번에 10페이지씩, 100페이지씩, 1000페이지씩 읽는 것과 같음 = 0.1, 0.01, 0.001
# 경사하강법에서 가중치와 편향을 수정할 때 얼마나 수정할지 결정하는 값, 학습률이 너무 크면 loss가 발산하고, 학습률이 너무 작으면 loss가 수렴하지 않음

# ++ 추가
# 유사역행렬 : 역행렬이 존재하지 않는 경우에도 계산이 가능하도록 일반화한 행렬, 선형회귀에서 가중치와 편향을 계산할 때 사용됨
# 유사역행렬회귀 : 선형회귀의 가중치와 편향을 계산하는 방법 중 하나, 입력값과 출력값의 관계를 수학적으로 계산하여 가중치와 편향을 구함
```
