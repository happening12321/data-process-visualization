```python
# 지도 학습: 입력 데이터와 정답 label을 함께 사용해 새로운 데이터의 정답을 예측하는 방법
# -> 과거 고객 이탈 데이터를 학습하여 새로운 고객의 이탈 여부를 예측

# 비지도 학습: 정답 label 없이 데이터 자체에 숨어 있는 구조나 패턴을 찾는 방법
# -> 고객 유형 분류(정답 x, 성별로 나이로 구매금액으로 주제별로 묶을 수 있음)

# 강화 학습: 보상체계에 따른 더 많은 보상을 학습하도록 하는 방법
# -> AI가 스스로 정답을 찾아가는 미로찾기

# 클러스터링(군집화): 서로 비슷한 특성을 가진 데이터를 그룹화
# -> 어질러진 물건을 색 또는 용도처럼 비슷한 특징에 따라 여러 상자에 나누어 담음

# k-means(k-평균): 데이터를 사용자가 미리 지정한 k개의 cluster 로 나누는 군집화 알고리즘
# -> 고객을 2그룹으로 나누고 싶다면 k=2로 설정한 뒤 두 중심점에 가까운 고객끼리 묶는다

# 중심점 : 케이민즈에서클리스터를대표하는가운데위치
#->에이 비 그룹 => 에이(방문, 1, 결제 10), 비(방문,2,결제,20)의
#중심점은 (1+2)/2+1.5와 (10+20)/2=15로 나뉠 수 있음

# 유클리드 거리 : 두 데이터가 숫자 공간에서 얼마나 떨어져 있는지를 재는 직선 거리

# 엘보 : k를 여러값으로 바꾸며 감소 폭을 비교해 적절한 k 
```

```python
# Dimension(차원): Feature 의 개수
# -> x가 공부시간 문제 풀이 수 -> 2차원, x가 공부시간, 문제풀이 수, 모의고사 -> 3차원

# 차원의 저주 : 차원이 증가하면서 데이터가 희소해지고, 거리, 학습, 시각화에 문제가 생기는 현상
# -> Feature가 2개인 경우 해석이 쉬웠는데 Feature가 500개가 된 경우 k-means 에서 거리가 일정하게 늘어나서 해석이 측정이 어려워질 수 있음

# noise: 분석 목적과 관계가 약하거나 우연히 섞인 불필요한 변동
# -> 학생 성적을 예측하는데  Feature에 사물함 번호가 들어가 있으며 의미없는 패턴을 학습할 수 있음

# 공분산 : 두 Feature가 함께 증가하거나 함께 감소하는 방향
# -> 공부 시간이 늘수록 문제 풀이 수도 늘어난다 -> 두 Feature에 공분산이 양수일 가능성이 높다

# PCA (주성분분석): 원본 데이터의 분산을 최대한 보존하는 새로운 축을 만들고 축 일부만 사용하여 차원을 줄인다
# - > 공부시간 문제풀이수 점수를 조합하여 전반적인 학습수준이라는 새로운 축을 생성한다.
# 데이터들이 뭘 말하고 싶은가를 모를 때 주성분 분석 때리면 특성 파악이 조금 되는 편

# 
```
