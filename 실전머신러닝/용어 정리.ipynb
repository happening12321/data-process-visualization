```python
# 편향: 학습 데이터를 바꿔 여러 번 생성한 모델의 평균적인 예측이 실제 관측에서 체계적으로 벗어나는 정도
# -> 실제 매출이 광고비에 따라 비선형 형태로 변하는데
# -> 선형 회귀를 사용해 직선 하나로만 예측을 한다고 가정해보면 고객 데이터를 바꿔도
# 비슷한 구간에서 지속적으로 틀린다면 편향을 고려해볼 수 있다

# 분산: 학습 데이터가 바꼈을 때 같은 입력에 대한 모델 예측이 달라지는 정도
# -> 성별, 목소리 -> 2가지 힌트를 가지고 DS반의 학생을 특정할 수 있다
# -> 성별, 목소리, AI 역량평가 1차, 2차, 키, 생일 -> 6가지 힌트를 가지고 DS반의 학생을 특정할 수 있다


# 과소적합: 모델이 학습 데이터의 패턴을 충분하게 파악하지 못함
# -> 보통 디시젼트리에서 트리 깊이를 1 정도로 해 충분한 패턴을 학습하지 못할 때 발생할 수 있다



# 행 ㅇ수와 컬럼을 늘려서 해결하고자 함

# 과적합: 모델이 학습의 우연한 패턴까지 학습해 새로운 데이터에 대한 적응력이 떨어짐
# 사탕: 포도, 망고, 콜라, 레몬
# 신 정도: 0.1, 0.4, 0.05, 0.7
# 모델이 0.1이 포도맛이라고 정답을 외워버리는 경우
# -> 새롭게 포도맛(0.15)이 추가되면 못맞히게 되어버리는 경우

# 노이즈: 패턴이 없이 우연한 변동이나 측정 오차가 있는 경우
# -> 평상 시에 구매가 없던 A고객이 갑자기 구매하는 경우
# -> 1월 중 하루 낮 온도가 20도까지 오른 경우

# 하이퍼파라미터 튜닝: 여러 설정값의 후보를 통해 좋은 성능을 찾는 과정
# -> epoch(학습 횟수): 1/10/ -> 5
# learning rate(학습률): 24/2.4->6
# 하이퍼파라미터 튜닝은 모델의 성능을 더 높이기 위해 학습률이나 학습횟수 등을 바꿔
# 결과를 더 높이는 최적의 모델을 찾으려는 과정

# 규제(regularization): 모델이 학습 데이터에 과적합되지 않도록 제약을 주는 것
# 사탕: 포도, 망고, 콜라, 레몬
# 신 정도: 0.1, 0.4, 0.05, 0.7
# 색: 보라, 주황, 검정, 노랑

# -> 모델이 0.1 


### 앙상블 : 여러 모델의 예측을 결합해 하나의 최종 예측을 만드는 방법
### 대표적인 앙상블 방법 : 트리형 모델(랜덤포레스트), 배깅

### 배깅 : 부트스트랩으로 만든 여러 표본에 각각 모델을 학습시키고 예측을 결합하는 앙상블 방법
### -> 고객 1,000명에서 여러 학습 표본을 만들고 각 표본으로 모델링을 학습해 예측 결과를 결합하는 것

### 랜덤포레스트 : 여러 의사결정 나무를 학습하고 각 분할에서 특성 후보를 무작위로 고르는 방식을 활용, 예측을 결합하는 모델
### -> 주택 가격을 예측할 때 나무 a는 면적을 중심으로 나무 b는 위치를 중심으로 분할할 수 있다

### OOB : 여러 결정트리를 만들 때 각 트리가 학습할 데이터를 중복을 허용해 무작위로 뽑는데 어떤 데이터들은 여러 번 뽑히고 어떤 데이터는 한 번도 안 뽑힌다면 -> 한 번도 안 뽑힌 데이터를 OOB라고 부른다

### OOB score : 1을 예측할 때 1을 학습하지 않은 데이터만 모아서 학습하는데 예측 결과와 실제 정답을 비교해 계산한 점수
### -> 학습하지 않은 데이터를 얼마나 잘 예측할 수 있는가
```
